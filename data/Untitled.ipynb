```python
import dadi
fs = dadi.Spectrum.from_file('data.fs')
print('shape   ', fs.shape)
print('folded  ', fs.folded)
print('pop_ids ', fs.pop_ids)
print('S()     ', fs.S())
print('masked  ', fs.mask.sum())
print('sum all ', fs.data.sum())
```

```python
import dadi, numpy as np
fs = dadi.Spectrum.from_file('data.fs')
idx = np.argwhere(fs.mask)
print('n masked:', len(idx))
print('corners :', fs.mask[0,0], fs.mask[-1,-1])
print('sum i+j over masked, min/max:', (idx.sum(1)).min(), (idx.sum(1)).max())
print('first few:', idx[:5].tolist())
print('last few :', idx[-5:].tolist())
```
